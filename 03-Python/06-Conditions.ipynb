##### Conditional Statements

1.  Conditional statements are those statements that makes certain decisions based on some conditions.

2.  Based on the conditions, they execute certain statements.

##### Types of Conditional Statements:

1.  if statement

if(condition):

    statements

```python
##### Check number is positive
num=10
if(num>0):
    print("Positive")
```

2.  if else statement

if(condition):

    statement1

else:

    statement2

```python
##### Check if the number is positive or negative
num = 50
if(num>0):
    print(f"{num} is positive")
else:
    print(f"{num} is negative")
```

3. elif statement

if condition1:

    statement1

elif condition2:

    statement2

else:

    statement3

```python
##### Check if the number is positive, negative or zero

num = 80
if(num>0):
    print(f"{num} is positive")
elif(num<0):
    print(f"{num} is negative")
else:
    print(f"{num} is zero")
```

4.   Nested if statements

if statement:

    if statement:

    statements

```python
age=18
citizen=True

##### Check if the person is eligible to vote
if(age>=18):
    if(citizen==True):
        print(f"The person is eligible to vote")
```

##### LOGICAL CONDITIONS

1.  and

Both conditions must be true.

```python
age = 18

salary = 35000

if((age>=18) and (salary>=35000)):
    print("Eligible")
```

2.  or

Either of the conditions is true

```python
marks = 95

if((marks >= 90) or (marks == 100)):
    print("Excellent")
```

3.  not

Reverses the condition

```python
flag = False

if(not(flag)):
    print("This statement is true")
```
