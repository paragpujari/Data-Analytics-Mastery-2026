##### Variable

    A Variable can be defined as a placeholder or a reserved memory location that is used to store the value in the memory.

    Each and every value of the variable has a memory that is referring to it.

##### Syntax of Variable

    var_name = value

```python
##### Define a variable
a = 10
print(a)
```

```python
##### Get the memory address of the variable
print(id(a))
```

##### OBSERVATIONS:

1.  The id function is used to get the memory address of the particular value in the memory.

##### Rules for defining the Variable

1.  A Variable can have letters, numbers or an underscore.

```python
x = 10
print(x)

x1 = 10
print(x1)

first_name = 'Parag'
print(first_name)
```

2. A Variable name can start with a letter or an underscore

```python
x = 10
print(x)

_x = 10
print(_x)
```

3. A variable name can have an underscore if needed

```python
first_name = 'Parag'
print(first_name)
```

4.  A Variable should not begin with a digit

```python
# 1name = 'Parag'  ##### Invlid variable :  A variable should not begin with a digit
```

5.  A variable name should not have spaces or hyphen in between

```python
# first name = 'Parag'  ##### Invalid variable :  A variable should not have spaces in between
# last-name  = 'Pujari' ##### Invalid variable :  A variable should not have hyphen in between
```

6.  Variables are case sensitive

```python
a = 10
A = 15

print(a)
print(A)
```

7.  A Variable name should not have @,$,% etc or any special symbols.

```python
# @name = 'Parag'     ##### Invalid variable :  A variable should not have @
# $name = 'Parag'     ##### Invalid variable :  A variable should not have $
# %name = 'Parag'     ##### Invalid variable :  A variable should not have %
```

```python
##### Naming convention --------->  A Variable must have a snake case
first_name = 'Parag'
print(first_name)

last_name  = 'Pujari'
print(last_name)
```

```python
##### Defining Multiple Variables with Multiple Values in one line
name, age, salary = 'Parag', 31, 53874

print(name)
print(age)
print(salary)
```

```python
##### Defining Multiple Variables with single Value in one line
x = y = z = 100

print(x)
print(y)
print(z)
```

```python
##### Reassignment
x = 50
print(f'Old value is {x}')

x = 80
print(f'New value is {x}')
```

```python
##### Swapping of the variables
a = 10
b = 20

print(f'Value of a before swapping is {a}')
print(f'Value of b before swapping is {b}')


a,b = b,a

print(f'Value of a after swapping is {a}')
print(f'Value of b after swapping is {b}')
```

```python
##### Constant -------> It is a value that cannot be changed during the entire execution of the program

PI = 3.142

print(f'Value of pi is {PI}')
```

```python
##### Memory address of the variable
x = 10
print(x)
print(id(x))

y = x
print(y)
print(id(y))
```

1.  y = x means that both x and y are referring to the same memory address.
