##### Type Conversions

Type Conversions refers to the tecnique of converting the value from one data type to another.

##### Types of Type Conversions

1.  Implicit Type Conversion ----->  It refers to the technique of automatically converting the value from one data type to another.

```python
x = 10
y = 5.6
print(x)
print(type(x))
print('--------------------------------------------------------')
res = x + y
print(res)
print(type(res))
```

##### OBSERVATIONS:

1.  Here the integer value is automatically converted into floating value by adding a floating value to it.

2.  The lower data type (int) is automatically converted into higher data type(float). 

3.  As the lower data type is converted to higher data type, so there is no loss of precision .

2.  Explicit Type Conversions -------> It refers to the technique of manually converting the value from one data type to another using function.

#####  Function	Conversion

1.  int()   --------->    	Converts a compatible value to an integer

2.  float()	--------->      Converts a compatible value to a float

3.  str()	--------->      Converts a value to a string

4.  bool()	--------->      Converts a value to a Boolean

5.  list()	---------->     Converts an iterable to a list

6.  tuple() ----------->    Converts an iterable to a tuple

7.  set()   ----------> 	Converts an iterable to a set

8.  complex() ------->  	Converts a compatible numeric value to a complex number
