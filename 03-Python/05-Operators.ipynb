##### Operators

1.  An operator can be defined as special symbols with which we can perform certain operations on the data.

2.  It is basically used for performing mathematical operations.

##### Main Operators Categories

##### Category	        Operators	                    Purpose

1.      Arithmetic  	+ - * / // % **	             Mathematical calculations

2.      Comparison  	== != > < >= <=	            Compare values

3.      Assignment	    = += -= *= /= //= %= **=	Assign/update values

4.      Logical	        and or not	                Combine conditions

5.      Bitwise	        & \| ^ ~ << >>	            Work at bit level

6.      Membership	    in, not in	                Check membership

7.      Identity    	is, is not              	Check object identity


##### Types of Operators

1.  Arithmetic Operators

```python
a = 10
b = 3

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)   ##### Removes the decimal parts after the point
print(a%b)
print(a**b)
```

2. Comparison Operators

```python
a = 10
b = 3

print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
print(a<b)
print(a>b)
```

3.  Assignment Operators

```python
a = 10
a+=3
print(a)

b = 10
b -= 3
print(b)

c = 10
c *= 3
print(c)

d = 10
d /= 3
print(d)

e = 10
e %= 3
print(e)

f = 10
f //= 3
print(f)

g = 10 
g **=3
print(g)
```

4.  Logical Operators

```python
print((10 > 5) and (8 > 6))

print((2 > 5) or (10 > 13))

print(not(10>5) or (9>8))
```

5.  Bitwise Operators

```python
a = 5
b = 3

##### perform bitwise AND
print(a&b)

##### perform bitwise OR
print(a|b)

##### perform bitwise XOR
print(a^b)

##### perform bitise NOT
print(~a)

##### perform botwise left shift
print(a<<1)

##### perform bitwise right shift
print(b>>2)
```

6.  Membership Operators

```python
print(5 in [1,2,3,4,5])
print(7 not in [1,3,5,7,9])
```

7. Identity Operator

```python
x = 10
print(id(x))

y = x
print(id(y))

print(x is y)
```

```python
x = [1,2,3]
print(id(x))
y = x
print(id(y))

print(x is y)
```

##### Operator Precedence

Python follows precedence rules.

```python
result = 10 + 5 * 2
print(result)
```

```python
result = (10 + 5) * 2
print(result)
```
