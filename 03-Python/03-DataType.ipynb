##### Data Type

1.  Data type defines the type that is assigned to each and every variable used in Python program.

2.  Here Data Type need not need to be declared automatically. It gets declared once the value is assigned to the variable.

```python
x = 10
print(x)
##### Determine the data type of x
print(type(x))
```

```python
name = 'Parag'
age  =  31
salary = 52874
height = 5.76
is_employed = True


print(type(name))
print(type(age))
print(type(salary))
print(type(height))
print(type(is_employed))
```

##### Data Types used in Python Program

1.  Integer --------->  These are the whole numbers without any decimal values

```python
x = 10
print(x)
print(type(x))
```

```python
age = 31
print(age)
print(type(age))
```

2.  Floating ---------> These are the values containing the decimal point

```python
height = 5.72
print(height)
print(type(height))
```

```python
marks = 87.5
print(marks)
print(type(marks))
```

3.  String ---------> It is a sequence of characters enclosed within single or double quotes

```python
name = 'Parag'

print(name)

print(type(name))
```

4.  Boolean -------> It represents two values. True or False

```python
is_employed = True

print(is_employed)

print(type(is_employed))
```

5. List ----------->  It is enclosed within []. It allows duplicate elements 

```python
marks = [80, 90, 80]
print(marks)
print(type(marks))
```

6. Tuples ---------> It is enclosed within (). It is immutable and allows duplicate elements

```python
marks = (80, 90, 80)
print(marks)
print(type(marks))
```

7. Sets -------->  It is enclosed within {}. It cannot have any duplicate elements in it

```python
elements = {10,20,30,40,20,30,50}

print(elements)

print(type(elements))
```

8.  Dictionary ----------> It is enclosed within {}. It allows key-value pairs in it.

```python
x = {'name':'parag','age':31,'salary':53874}

print(x)

print(type(x))
```

9.  None ---------> It represents the absence of a value.

```python
x = None
print(x)
print(type(x))
```

8.  What is type conversion? Explain int(), float() and str() with examples


Type conversion is a technique of converting the value of one type to another type.

Converting from lower to higher type results in no loss of precision and is safe. Ex:--  int -----> float

Converting from higher to lower type results in loss of precision and is dangerous. Ex:-- float ---> int

```python
x = 10
# int -----> float
ans = float(x)
print(ans)
```

```python
x = 10.23
# float -----> int
ans = int(x)
print(ans)
```

```python
x = 1200
# int -------> str
ans = str(x)
print(ans)
```
