##### Loops

1.  A Loop is a condition that is used to repeat the set of iterations unless and until a certain condition is satisfied.

##### Types of Loops

(a.)    for loop

(b.)    while loop

(a.)    for loop

```python
##### print the list of numbers from 1 till 10
for i in range(1,11):
    print(i)
```

```python
##### print the list of numbers using while loop
i = 1
while(i <= 10):
    print(i)
    i = i+1
```

22. Check whether the number is prime or not

```python
num = 13
flag=True

i=2
while(i<num):
    if(num%i==0):
        flag=False
        break
    i=i+1
if(flag):
    print(f"{num} is prime")
else:
    print(f"{num} is not prime")
```

24.  Check whether the number is palindrome or not

```python
num = 121
x = num
rev = 0
while(x!=0):
    d = x%10
    rev = (rev*10)+d 
    x = x//10
if(rev==num):
    print(f"{num} is palindrome")
else:
    print(f"{num} is not palindrome")
```

29. Check whether the number is armstrong or not

```python
n=153
x=n
s=0
while(x!=0):
    d=x%10
    s=s+(d**3)
    x=x//10
if(s==n):
    print(f"{n} is an armstrong number")
else:
    print(f"{n} is not an armstrong number")
```
