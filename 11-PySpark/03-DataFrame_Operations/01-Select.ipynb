**Select() statement**

1.  Select statement is one of the most important operation in DataFrame that helps in displaying the data from the original dataframe by selecting certain columns.

**Functions of Select() statement**

1.  Select () statement helps in displaying the data from the original dataframe.

2.  It helps in creating a new dataframe with the selected columns.

3.  It helps in only displaying the new dataframe but we cannot do any modifications on the selected dataframe.

4.  All the columns in the dataframe can be accessed using col() function.

**Creation of Spark Session**

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
##### Create a spark session
session = SparkSession.builder.appName("Select").getOrCreate()
##### Get the version of Spark Session
print(session.version)
```

**Creation of Employee DataFrame**

```python
data = [
    (101,	'Amit',	'IT',	50000)  ,
    (102,	'Priya','HR',	40000)  ,
    (103,	'Rahul','Finance',60000),
    (104,	'Neha' ,'IT',	55000),
    (105,	'Ravi','HR'	,45000)
]

columns = ["id", "name", "department","salary"]

##### Creation of the dataframe
emp_df = session.createDataFrame(data,columns)

##### Display the dataframe
emp_df.show()
```

**Five solved examples**

1.  Display only employee names from the above dataframe

```python
emp_df.select('name').show()
```

2. Display employee ID and name

```python
emp_df.select('id','name').show()
```

3. Display employee name, department, and salary

```python
emp_df.select('name','department','salary').show()
```

4.  Display employee ID and salary using the col() function

```python
emp_df.select(col('id'), col('salary')).show()
```

5. Display all columns using select()

```python
emp_df.select('*').show()
```
